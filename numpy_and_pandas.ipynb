**list_style vs dict_style**

```python
# Nested list style practice is list, and we use seperate column to give a heading to columns
parct=[ [1,"pak","south asia",2.50,],[2,"sharja","middle_east",1.0,],[3,"missisaga","Canada",5.50,]]
columns=['s.ro','continent','city_n','pop']
```

```python
df=pd.DataFrame(data=parct,columns=columns)
print(df)
```

```python
#dictionary style is used with index, no column header needed
consumers={  "Rank":["11", "12", "13"],
    "customer":["c1", "c2", "c3"],
           "purchaser":['y','n','n'],
           "medium":[20,40,50]
          }


df=pd.DataFrame(data=consumers,index=consumers['Rank'])
print(df)
```

**Pandas.Concat function** in series as list and in dataFrame

```python
import pandas as pd
tuesday=pd.Series({'lap1':'1:30','lap2':'1:35','lap3':'1:20','lap4':'1:40'})
wednesday=pd.Series({'lap1':'1:20','lap2':'1:15','lap3':'1:30','lap4':'1:20'})
laps=pd.concat([tuesday,wednesday],keys=['tuesday','wednesday'])
print(laps)
```

```python
import pandas as pd
df=pd.read_excel(r"/content/Customer Call List.xlsx")
```

```python
df
```

```python
df=df.drop_duplicates()
```

```python
df=df.drop(columns="Not_Useful_Column")
```

```python
df["Last_Name"]=df["Last_Name"].str.lstrip('/')
df["Last_Name"]=df["Last_Name"].str.lstrip('...')
df["Last_Name"]=df["Last_Name"].str.rstrip('_')
#df['Last_Name'].str.strip("_.../")
df
```

```python
df["Phone_Number"]=df["Phone_Number"].str.replace('[^a-zA-Z0-9]','')
```

```python
#df["Phone_Number"].apply(lambda x: x[0:3]+'-'+ x[3:6]+'-'+ x[6:10])
#df["Phone_Number"]=df["Phone_Number"].apply(lambda x: str(x))
#str(df['Phone_Number'])
#df["Phone_Number"].apply(lambda x: str(x))
#df["Phone_Number"].apply(lambda x: x[0:3]+ '-' + x[3:6]+ '-' + x[6:10])
df["Phone_Number"] = (
    df["Phone_Number"]
    .astype(str)
    .str.replace(r"\.0$", "", regex=True)
    .str.slice(0,3) + "-" +
    df["Phone_Number"].astype(str).str.slice(3,6) + "-" +
    df["Phone_Number"].astype(str).str.slice(6,10)
)

```
